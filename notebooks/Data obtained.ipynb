## Dataset Statistics

The code below generates descriptive statistics for the dataset.
It was obtained with AI assistance and adapted for this project.

The code produces:
- Data types for each column
- Missing value percentages
- Unique value counts
- Distribution by culture, country, period, and object type
- Output saved as a CSV file in the data/ folder

```python
import os

# Check current working directory
print("Current folder:", os.getcwd())

# List all files here
print("\nFiles in current folder:")
for f in os.listdir():
    print(f)
```

```python
import pandas as pd
import os

# Correct path — go up one level then into data folder
filename = "../data/asian_islamic_neareastern_met.csv"

df = pd.read_csv(filename, low_memory=False)

print("=== DATASET STATISTICS ===")
print(f"Rows: {len(df)}")
print(f"Columns: {len(df.columns)}")

print("\n--- Missing Values (%) ---")
print(df.isna().mean().round(2))

print("\n--- Unique Values per Column ---")
print(df.nunique())

print("\n--- Top Cultures ---")
print(df['Culture'].value_counts().head(10))

print("\n--- Top Countries ---")
print(df['Country'].value_counts().head(10))

print("\n--- Top Periods ---")
print(df['Period'].value_counts().head(10))

print("\n--- Object Types ---")
print(df['Object Name'].value_counts().head(10))

print("\n--- Public Domain ---")
print(df['Is Public Domain'].value_counts())

# Save statistics to data folder
missing_df = df.isna().mean().round(2).reset_index()
missing_df.columns = ['column', 'missing_percentage']
missing_df.to_csv('../data/dataset_statistics.csv', index=False)
print("\nStatistics saved to data/dataset_statistics.csv")
```
